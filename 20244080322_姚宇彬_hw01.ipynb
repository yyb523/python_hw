## 欧几里得算法
1. 实现`gcd(m,n)`函数，返回两个整数的最大公约数，无需错误检查
2. 测试三组：(2018, 2019)、(1200, 300)、(5040, 60)
3. 思考：传入负数参数，观察函数行为，分析是否合理


```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


# 测试用例
print("gcd(2018,2019) =", gcd(2018, 2019))
print("gcd(1200,300) =", gcd(1200, 300))
print("gcd(5040,60) =", gcd(5040, 60))

# 负数测试
print("\n负数参数测试 gcd(-1200,300) =", gcd(-1200, 300))
```

## 近似欧拉数 e
公式：
1. 极限形式：$e=\lim\limits_{x \to \infty}(1+\frac1x)^x$
2. 无穷级数：$e=\sum_{k=0}^{\infty}\frac{1}{k!}$，约定 $0! = 1$

函数需求：
1. `euler_limit(n)`：$(1+\frac1n)^n$ 近似e
2. `euler_infinite_sum(n)`：前n项级数求和；n=0返回0；n=1返回1；n=4返回 $1+1+\frac12+\frac16$
3. `euler_approx(epsilon)`：while循环，直到和与真实`math.exp(1)`误差小于$\epsilon$，返回近似e
4. `print_euler_sum_table(n)`：k从1~n，打印每一项`euler_infinite_sum(k)`



```python
import math


def euler_limit(n):
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 阶乘，0! = 1
    for k in range(n):
        if k == 0:
            term = 1 / fact
        else:
            fact = fact * k
            term = 1.0 / fact
        total += term
    return total


def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) >= epsilon:
        if k == 0:
            term = 1 / fact
        else:
            fact *= k
            term = 1.0 / fact
        total += term
        k += 1
    return total


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(f"k={k:2d} | sum = {val:.6f}")


# --------测试--------
print("euler_limit(10000):", euler_limit(10000))
print("真实e math.exp(1):", math.exp(1))

print("\neuler_infinite_sum(0):", euler_infinite_sum(0))
print("euler_infinite_sum(1):", euler_infinite_sum(1))
print("euler_infinite_sum(4):", euler_infinite_sum(4))

print("\neuler_approx(0.001):", euler_approx(0.001))

print("\n-----print_euler_sum_table(6)-----")
print_euler_sum_table(6)
```

## 测试整数属性
> 不允许使用`math.sqrt`，只允许除法、取模`%`
1. `is_power_of_2(n)`：输入正整数n，返回布尔值，判断n是否是2的幂 $n=2^p,\ p\ge0$
2. `is_power(b, n)`：泛化函数，输入b、n正整数，判断n是否是b的幂



```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True


def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True


# 测试
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(7):", is_power_of_2(7))
print("is_power_of_2(1):", is_power_of_2(1))

print("\nis_power(3,81):", is_power(3, 81))
print("is_power(3,80):", is_power(3, 80))
print("is_power(5,1):", is_power(5, 1))
```

### 任务1：score_level(score) 成绩等级函数
90~100优秀；80~不足90良好；60~不足80及格；0~不足60不及格；超出0‑100抛出ValueError。


```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0‑100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 测试
print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))

# 测试越界，取消注释看异常
# print(score_level(105))
```

### 任务3：迭代版、递归版阶乘函数，0! = 1


```python
# 迭代版阶乘
def fact_iter(n):
    if n < 0:
        raise ValueError("输入为非负整数")
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归版阶乘
def fact_recur(n):
    if n < 0:
        raise ValueError("输入为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n-1)

for num in [0, 1, 5, 10]:
    print(f"n={num},迭代阶乘:{fact_iter(num)},递归阶乘:{fact_recur(num)}")
```

### 任务4：summary(values) 返回最小值、最大值、平均值，元组返回


```python
def summary(values):
    if len(values) == 0:
        raise ValueError("列表不能为空")
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return min_v, max_v, avg_v

lst = [10,20,30,40,50]
min_val, max_val, avg_val = summary(lst)
print(f"最小值:{min_val},最大值:{max_val},平均值:{avg_val}")
```

### 任务3：parse_student(line)函数
将成绩转为整数，检查字段数、姓名非空、成绩0‑100；无效记录抛出ValueError


```python
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，分割得到{len(parts)}个字段")
    name_str, score_str = parts
    name_str = name_str.strip()
    score_str = score_str.strip()
    if len(name_str) == 0:
        raise ValueError("姓名为空")
    score = int(score_str)
    if not (0 <= score <= 100):
        raise ValueError(f"成绩{score}不在0‑100范围")
    return {"name": name_str, "score": score}

# 测试1：合法数据
res1 = parse_student("Alice,85")
print("测试1 合法记录：", res1)

# 测试2：成绩超出范围
try:
    parse_student("Henry,105")
except ValueError as e:
    print("测试2 异常：", e)

# 测试3：字段数量不对
try:
    parse_student("Eva")
except ValueError as e:
    print("测试3 异常：", e)

# 测试4：空名字
try:
    parse_student(",77")
except ValueError as e:
    print("测试4 异常：", e)
```

### 任务4：读取文件，捕获异常，记录行号、错误，有效记录保存字典列表


```python
import os

data_dir = os.path.join(os.getcwd(), "exp01_data")
file_path = os.path.join(data_dir, "students.txt")

valid_records = []
error_log = []

with open(file_path, mode="r", encoding="utf‑8") as f:
    lines = f.readlines()

for row_num, raw_line in enumerate(lines[1:], start=2):
    try:
        rec = parse_student(raw_line)
        valid_records.append(rec)
    except ValueError as err:
        error_log.append({"行号": row_num, "错误原因": str(err)})

print("✅有效记录：", valid_records)
print("\n❌错误记录：")
for err_item in error_log:
    print(err_item)
```

### 任务5：统计有效、无效记录数、平均分、及格率写入 report.txt；演示 w 与 a模式


```python
import os

# 统计
valid_count = len(valid_records)
invalid_count = len(error_log)
scores_valid = [r["score"] for r in valid_records]
avg_score = sum(scores_valid)/len(scores_valid)
pass_num = sum(1 for r in valid_records if r["score"] >= 60)
pass_rate = pass_num / valid_count

report_path = os.path.join(data_dir, "report.txt")

# w模式：覆盖写，会清空原有文件
with open(report_path, "w", encoding="utf‑8") as f:
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

# a模式：追加，在文件末尾新增内容，不清空原有
with open(report_path, "a", encoding="utf‑8") as f:
    f.write("---追加测试：a模式写入的这一行---\n")

# 读取查看report内容
with open(report_path, "r", encoding="utf‑8") as f:
    content = f.read()
print(content)
```


# ## 字符串的乐趣
# 1. is_palindrome 判断回文，忽略空格、大小写


```python
# %%
def is_palindrome(s: str) -> bool:
    # 去除空格，全部转为小写
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]


# 测试
print(is_palindrome("T A C O cat"))   # True
print(is_palindrome("level"))         # True
print(is_palindrome("hello"))         # False
```


# 2. is_abecedarian 判断字母是否按字母升序，忽略空格


```python
# %%
def is_abecedarian(s: str) -> bool:
    cleaned = s.replace(" ", "").lower()
    # 遍历，比较相邻字符
    for i in range(len(cleaned)-1):
        if cleaned[i] > cleaned[i+1]:
            return False
    return True


# 测试
print(is_abecedarian("adder"))          # True
print(is_abecedarian("abcd efgh xyz"))  # True
print(is_abecedarian("dog"))            # False
```


# 3. remove_vowels 删除元音 a e i o u


```python
# %%
def remove_vowels(s: str) -> str:
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)


# 测试
print(remove_vowels('cat'))     # ct
print(remove_vowels('audio'))   # d
print(remove_vowels("aeiou"))   # ""
```


# ## 列表的乐趣
# 1. list_reverse 反转列表，非列表抛异常


```python
# %%
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]


# 测试
print(list_reverse([1,2,3]))   # [3,2,1]
# list_reverse((1,2,3)) # 会抛出TypeError
```


# 2. is_sorted 判断序列升序，只遍历一次


```python
# %%
def is_sorted(seq) -> bool:
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


# 测试
print(is_sorted([1,2,3,4]))     # True
print(is_sorted([1,3,2]))       # False
print(is_sorted((2,2,3,5)))     # True
```


# ## 作为向量的元组
# 1. vec_scalar_mult 向量元组标量乘法


```python
# %%
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须为整数或浮点数")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    return tuple(x * s for x in t)


# 测试
print(vec_scalar_mult((1,2,3), 2))  # (2,4,6)
```


# 2. vec_inner_product 向量内积


```python
# %%
def vec_inner_product(x, y) -> float:
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量元组长度必须相等")
    total = 0.0
    for xi, yi in zip(x, y):
        total += xi * yi
    return total


# 测试
print(vec_inner_product((1,2), (3,4)))  # 1*3 +2*4 =11.0
```

# 3. check_valid_mx 检查是否合法矩阵（元组的元组）


```python
# %%
def check_valid_mx(mx) -> bool:
    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 空矩阵处理
    if len(mx) == 0:
        return True
    # 所有行长度一致
    row_len = len(mx[0])
    for row in mx:
        if len(row) != row_len:
            return False
        # 每个元素必须数字
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


# 测试
mx1 = ((1,2), (3,4))
print(check_valid_mx(mx1)) # True
```


# 4. mx_vec_mult 矩阵乘向量


```python
# %%
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) == 0:
        raise ValueError("矩阵不能为空")
    n_col = len(mx[0])
    if len(vec) != n_col:
        raise ValueError("矩阵列数和向量维度不一致")
    result = []
    for row in mx:
        val = vec_inner_product(row, vec)
        result.append(val)
    return tuple(result)


# 测试
M = ((1,2), (3,4))
v = (5,6)
print(mx_vec_mult(M, v)) # (17.0, 39.0)
```


# ## 更多向量的乐趣：Vector 类
#
# **思考题：维度dim和元素elements选择实例属性**
#
# 理由：
# 每个Vector实例都拥有属于自己独立的维度与元素数据。
# 如果设置为类属性，则全部Vector对象会共享同一套dim与elements；修改任意一个实例会影响全部实例，不符合向量对象的语义。
# 因此应当作为实例属性，在`__init__`内部绑定给`self`。


```python
# %%
from functools import total_ordering

@total_ordering
class Vector:
    def __init__(self, dim, elements=None):
        # 校验维度
        if not isinstance(dim, int):
            raise TypeError("维度必须为整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        # elements默认，不给则全部0
        if elements is None:
            elements = [0.0]*dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("elements必须是list或tuple")
        # 校验元素全部数字
        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("向量元素必须是数字")
        if len(elements) != dim:
            raise ValueError(f"维度{dim}和元素数量{len(elements)}不匹配")

        # =====实例属性=====
        self.dim = dim
        self.elements = tuple(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector实例比较")
        if self.dim != other.dim:
            raise ValueError("比较的两个Vector维度不同")
        return self.elements < other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致，无法计算内积")
        s = 0.0
        for a,b in zip(self.elements, other.elements):
            s += a*b
        return s

    def __mul__(self, other):
        """ v * x : 向量右乘，other可以标量或者Vector"""
        if isinstance(other, (int, float)):
            new_elems = [x*other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量相乘维度不一致")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("*操作只支持标量(int/float)或Vector")

    def __rmul__(self, scalar):
        """ scalar * v  标量左乘 """
        # 直接复用 __mul__
        return self * scalar

    def __repr__(self):
        return f"Vector({self.dim}, {list(self.elements)})"


# ========== Vector测试 ==========
v1 = Vector(2, (2,3))
v2 = Vector(2, (4,5))
print(v1)
print(v2)
print(v1 == Vector(2,[2,3]))
print(v1 < v2)
print(v1.dot(v2))
print(v1 * v2)      # 逐元素相乘 Vector(2, [8.0,15.0])
print(2 * v1)       # 标量左乘
print(v1 * 3)       # 标量右乘
```

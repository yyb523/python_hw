## 实验1：正则表达式基础匹配
测试 \d+ 、\s+ 、\bcat\b；交替匹配；量词优先级；* + ? {m} {m,n}
> 全部打印匹配结果用于验证


```python
import hashlib
import os
import platform
import uuid
import sys
import re
from bs4 import BeautifulSoup
s1 = "cat on mat; later: cats and dogs"
print("源文本1：", s1)
print("\\d+匹配结果:", re.findall(r"\d+", s1))
print("\\s+匹配结果:", re.findall(r"\s+", s1))
print(r"\bcat\b匹配结果:", re.findall(r"\bcat\b", s1))

s2 = "cat dog bird CAT"
print("\n源文本2：", s2)
print("cat|dog匹配结果 :", re.findall(r"cat|dog", s2))

print('\nre.findall(r"a|aa|aaa", "aaaa") :', re.findall(r"a|aa|aaa", "aaaa"))
print('re.findall(r"aaa|aa|a", "aaaa") :', re.findall(r"aaa|aa|a", "aaaa"))

s3 = "ct cat caat caaat"
print("\n源文本3：", s3)
print("ca*t :", re.findall(r"ca*t", s3))
print("ca+t :", re.findall(r"ca+t", s3))
print("ca?t :", re.findall(r"ca?t", s3))
print("ca{2}t :", re.findall(r"ca{2}t", s3))
print("ca{1,2}t :", re.findall(r"ca{1,2}t", s3))
```

## 实验2：分组、回溯引用与替换
1.捕获组提取邮箱用户名域名
2.后向引用匹配连续重复单词
3.日期格式转换：年-月-日 → 月-日-年，兼容 - / . 分隔符
4.电话号码统一格式为 (123) 456‑7890
> 打印原始与处理后文本对比验证


```python
#1提取邮箱用户名域名
mail_text = "联系:alice@example.com;备选:bob@test.org"
mail_pat = r"([a-zA-Z0-9]+)@([a-zA-Z0-9.]+)"
mails = re.findall(mail_pat, mail_text)
print("原始文本：", mail_text)
print("邮箱(用户名,域名):", mails)

#2后向引用连续重复单词
rep_text = "cat cat, dog cat, bird bird"
rep_res = re.findall(r"\b(\w+)\b \1", rep_text)
print("\n原始文本：", rep_text)
print("重复单词：", rep_res)

#3日期转换
date_text = """2024-05-01
2024/05/02
2024.05.03"""
print("\n====日期转换前====")
print(date_text)
date_pat = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
date_new = re.sub(date_pat, r"\2-\3-\1", date_text)
print("====日期转换后====")
print(date_new)

#4电话号码格式化
phone_text = "联系我:123-456-7890 或 987.654.3210"
phone_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
phone_new = re.sub(phone_pat, r"(\1) \2-\3", phone_text)
print("\n原始电话文本：", phone_text)
print("电话转换结果：", phone_new)
```

## 实验3：BeautifulSoup解析HTML
解析给定html字符串；获取标题、h1；find_all筛选class和a标签；对比不同对象get_text()输出
> 逐项打印输出验证解析结果


```python
html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
#任务1
title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name, "h1文本：", h1_tag.get_text())

#任务2
p_course = soup.find_all("p", class_="course")
print("\nclass=course的p段落：")
for idx,p in enumerate(p_course):
    print(f"p[{idx}]：", p.get_text())

links = soup.find_all("a")
print("\n所有链接：")
for idx,a in enumerate(links):
    print(f"a[{idx}] 链接文字:{a.get_text()}, href={a['href']}")

#任务3 get_text对比
print("\n====全部文档get_text====")
print(soup.get_text())
print("\n====单个p段落get_text====")
print(p_course[0].get_text())
print("\n====单个a链接get_text====")
print(links[0].get_text())
```
